```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
text = """Good morning everyone,

Today, I would like to talk about the importance of learning. Learning is a continuous process that helps us grow personally and professionally. It is not limited to schools, colleges, or textbooks. We learn from our experiences, our mistakes, other people, and even from our failures.

In today's rapidly changing world, learning new skills has become more important than ever. Technology is developing quickly, and new tools and opportunities appear every day. As students and future professionals, we need to keep improving ourselves and adapting to these changes.

I believe that we should focus not only on getting good marks or a good job, but also on developing the ability to solve problems and learn independently. When we face a difficult situation, instead of being afraid of failure, we should treat it as an opportunity to learn something new.

Another important part of learning is consistency. We do not need to learn everything in one day. Even spending thirty minutes every day learning something useful can make a big difference over time.

In conclusion, learning makes us more confident, capable, and prepared for the future. So, we should always remain curious, accept our mistakes, and continue learning throughout our lives.

Thank you."""

```

```python
splitter = RecursiveCharacterTextSplitter(chunk_size = 50, chunk_overlap = 20)
```

```python
chunks = splitter.split_text(text)
```

#### Documents Text Loader

```python
from langchain_community.document_loaders import TextLoader, PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
loader = TextLoader("../utils/speech.txt")
```

```python
text = loader.load()
```

```python
pdfloader = PyPDFLoader("../utils/MLFundamentals.pdf")
pdfs =pdfloader.load()
```

```python
splitter = RecursiveCharacterTextSplitter(chunk_size = 1000, chunk_overlap = 200)
```

```python
text_chunks = splitter.split_documents(text)
len(text_chunks)
```

```python
pdf_chunks = splitter.split_documents(pdfs)
len(pdf_chunks)
```
