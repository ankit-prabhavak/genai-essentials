```python
from pydantic import BaseModel
```

```python
class User(BaseModel):
    name : str
    age : int
```

```python
user = User(name="Ankit", age=26)

user
```

#### Field Validation

```python
from pydantic import Field
from typing import List, Annotated
```

```python
class Product(BaseModel):
    title : str = Field(max_length=20)
    des : str = Field(min_length=20)
    price : int = Field(gt=0, le=100)
```

```python
product = Product(title="Samsung S23", des="Best in the market for young professionals", price=75)
```

```python
product
```

```python
class Order(BaseModel):
    id : str = Field(min_length=6, max_length=15)
    product : Product
```

```python
order = Order(id="jana12", product=product)
order
```

```python
class Carts(BaseModel):
    id : str = Field(min_length=6, max_length=15)
    items : List[Product]
    comment: str = Field(default="No comment")
```

```python
cart = Carts(id="dhjfdh", items=[product], comment="Kya Mast Hai")
```

```python
cart
```

```python
# Annotated is used to make common validator

common_validator = Annotated[List[Product], "This is common validator"]
```

```python
class Order(BaseModel):
    id : str = Field(min_length=6, max_length=15)
    product : common_validator
```
