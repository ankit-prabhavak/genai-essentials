#### Loading Data From Wikipedia

```python
import wikipedia
from langchain_community.document_loaders import WikipediaLoader

wikipedia.set_user_agent("MyRAGApp/1.0 (contact@yourdomain.com)")

loader = WikipediaLoader(query="samsung", load_max_docs=2)
raw_documents = loader.load()
print(f"Loaded {len(raw_documents)} Wikipedia page(s).")
```
