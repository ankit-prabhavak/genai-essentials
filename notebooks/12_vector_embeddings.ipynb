```python
from dotenv import load_dotenv
import os
```

```python
load_dotenv()
```

```python
documents = [
    "Artificial Intelligence is the field of creating machines that can perform tasks requiring human intelligence.",
    
    "Machine Learning is a subset of Artificial Intelligence where systems learn patterns from data.",
    
    "Deep Learning uses neural networks with multiple layers to learn complex patterns from large datasets.",
    
    "Natural Language Processing allows computers to understand and process human language.",
    
    "Embeddings convert text into numerical vectors that capture the meaning and semantic relationships between words or sentences.",
    
    "Vector databases store embeddings and allow applications to perform similarity searches.",
    
    "Retrieval-Augmented Generation combines document retrieval with a language model to generate answers using external knowledge.",
    
    "LangChain is a framework for building applications powered by language models.",
    
    "LangGraph is a framework for building stateful, multi-step AI agents and workflows.",
    
    "Groq provides fast inference for large language models."
]
```

```python
from langchain_openai import OpenAIEmbeddings
```

```python
embeddings = OpenAIEmbeddings(
    model="text-embedding-3-small",
    api_key=os.getenv("OPENROUTER_API_KEY"),
    base_url="https://openrouter.ai/api/v1"
)
```

```python
vector = embeddings.embed_documents(documents)
```

```python
len(vector[0])
```

#### Store Vector Embedding in Vector DB (Chroma)

```python
from langchain_chroma import Chroma
```

```python
vector_store = Chroma.from_texts(
    texts=documents,
    embedding=embeddings,
    persist_directory="./vector_db"
)
```

```python
query = "Natural Language Processing"
```

```python
result = vector_store.similarity_search(query=query , k = 2)  # k denotes number of similar results
result
```
