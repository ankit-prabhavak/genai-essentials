```python
import os
```

```python
from langchain_openai import ChatOpenAI
llm = ChatOpenAI(
    model="openai/gpt-oss-20b:free",
    api_key=os.getenv("OPENROUTER_API_KEY"),
    base_url="https://openrouter.ai/api/v1"
)
```

#### How to write prompts in standard way using dictionaries

```python
prompts = [
    {"role": "system", "content": "you are a python developer"}, # "system", "ai", "user"
    {"role": "user", "content": "write a function to add two numbers"},
]

res = llm.invoke(prompts)
print(res.content)
```

```python
prompts = [
    {"role": "system", "content": "you are a translator that translates english text to hindi"}, # "system", "ai", "user"
    {"role": "user", "content": "i love to play cricket"},
]

res = llm.invoke(prompts)
print(res.content)
```

#### Prompts Chaining | Prompt Templates or Dynamic Prompts

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

out = StrOutputParser()

prompt = ChatPromptTemplate.from_messages([
    {"role": "system", "content": "you are a translator that translates english text to {language}"}, # "system", "ai", "user"
    {"role": "user", "content": "{text}"},
])

```

```python
final_prompt = prompt.format_prompt(language="hindi", text="i love to play cricket")

res = llm.invoke(final_prompt)
print(res.content)
```

#### Runnable (Chains)

A Runnable in **LangChain** is a standardized component that can receive input, perform an operation, and produce output, allowing different components such as prompts, LLMs, retrievers, and parsers to be composed into chains.

```python
# prompt.invoke({"language": "hindi", "text": "i love to play cricket"})

# llm.invoke()

# chaining
res = llm.invoke(prompt.invoke({"language": "hindi", "text": "i love to play cricket"}))
print(res.content)
```

#### Custom text transformer for chaining

```python
def transform_case(text: str) -> str:
    return text.upper()
```

```python
chains = prompt | llm | out | transform_case

res = chains.invoke({"language": "German", "text": "I love to work with LangChain"})
# print(res.content) not needed as the output parser returns a string
print(res)
```
